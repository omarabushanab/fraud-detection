```python
import smtplib
from email.message import EmailMessage

def send_test_email(subject, body, is_html=False, attachment_path=None):
    # Setup connection
    sender_email = "omarabushanab03@gmail.com"
    receiver_email = "abushanab195@gmail.com"
    # Note: Use a Google App Password, NOT your regular password
    password = "mtst zhwi nxwy abnx" 

    msg = EmailMessage()
    msg['Subject'] = subject
    msg['From'] = sender_email
    msg['To'] = receiver_email

    if is_html:
        msg.set_content("This is a fallback plain-text version.")
        msg.add_alternative(body, subtype='html')
    else:
        msg.set_content(body)

    # Adding attachments if needed
    if attachment_path:
        with open(attachment_path, 'rb') as f:
            file_data = f.read()
            file_name = f.name.split('/')[-1]
        msg.add_attachment(file_data, maintype='application', subtype='octet-stream', filename=file_name)

    # Send via Gmail SMTP
    try:
        with smtplib.SMTP_SSL('smtp.gmail.com', 465) as smtp:
            smtp.login(sender_email, password)
            smtp.send_message(msg)
        print(f"Success: {subject} sent.")
    except Exception as e:
        print(f"Error: {e}")

# --- Example Usage for Your Matrix ---
# 1. Arabic Phishing (Plain Text)
send_test_email(
    "تحديث أمني هام", 
    "يرجى النقر هنا لتحديث بياناتك البنكية فوراً: http://fake-bank-login.com", 
    is_html=False
)

# 2. English Phishing (HTML + Attachment)
html_body = "<h3>Urgent: Invoice Overdue</h3><p>Please check the attached <a href='http://malicious-link.com'>Invoice</a>.</p>"
send_test_email("Overdue Invoice #9921", html_body, is_html=True, attachment_path="invoice.pdf")
```
