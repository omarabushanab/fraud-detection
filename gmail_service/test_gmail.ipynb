```python
import smtplib
from email.message import EmailMessage

# --- CONFIGURATION ---
SMTP_SERVER = "smtp.gmail.com"
SMTP_PORT = 465
SENDER_EMAIL = "your-test-sender@gmail.com"
SENDER_PASSWORD = "your-app-password" # Not your login password!
TARGET_EMAIL = "the-email-you-are-monitoring@gmail.com"

def send_test_email(subject, content, attachment_name=None, attachment_data=None):
    msg = EmailMessage()
    msg['Subject'] = subject
    msg['From'] = SENDER_EMAIL
    msg['To'] = TARGET_EMAIL
    msg.set_content(content)

    if attachment_name and attachment_data:
        msg.add_attachment(
            attachment_data,
            maintype='application',
            subtype='octet-stream',
            filename=attachment_name
        )

    with smtplib.SMTP_SSL(SMTP_SERVER, SMTP_PORT) as smtp:
        smtp.login(SENDER_EMAIL, SENDER_PASSWORD)
        smtp.send_message(msg)
    print(f"✅ Sent: {subject}")

# --- TEST CASES ---

# 1. URL Phishing Test (Triggers URL Classifier)
send_test_email(
    "Urgent: Account Verification",
    "Please verify your account at http://phishing-portal.com/login"
)

# 2. XLM-R Model Test (Triggers Text Analysis)
send_test_email(
    "Winner Notification",
    "You have won a lottery prize of $1,000,000. Send your bank details immediately to claim."
)

# 3. Attachment Scan Test (Triggers ClamAV/VBA)
send_test_email(
    "Invoices Attached",
    "Please see the attached document for your monthly bill.",
    attachment_name="invoice.txt",
    attachment_data=b"X5O!P%@AP[4\\PZX54(P^)7CC)7}$EICAR-STANDARD-ANTIVIRUS-TEST-FILE!$H+H*" # Standard EICAR test string
)
```
